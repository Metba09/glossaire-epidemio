```python

from google.colab import files
import os

# 1. Ouvre le sélecteur de fichiers de votre ordinateur
print("👉 Veuillez sélectionner votre fichier ZIP dans la fenêtre qui va apparaître...")
uploaded = files.upload()

# 2. Récupère le nom exact du fichier que vous venez d'uploader
nom_du_zip = list(uploaded.keys())[0]
print(f"\n✅ Fichier reçu : {nom_du_zip}")

# 3. Décompresse le ZIP (le -o écrase les fichiers s'ils existent déjà)
print("\n⏳ Décompression en cours...")
!unzip -q -o "$nom_du_zip"
print("✅ Décompression terminée.")

# 4. Trouve automatiquement le dossier qui contient le projet Remotion (package.json)
dossier_projet = None
for root, dirs, files_list in os.walk('.'):
    if 'package.json' in files_list:
        dossier_projet = root
        break

if dossier_projet and dossier_projet != '.':
    print(f"\n📂 Projet Remotion trouvé dans le sous-dossier : {dossier_projet}")
    os.chdir(dossier_projet) # On entre dans le dossier
elif 'package.json' in os.listdir('.'):
    print("\n✅ Les fichiers du projet sont directement à la racine.")
else:
    print("\n⚠️ ATTENTION : Aucun fichier 'package.json' trouvé. Votre ZIP n'est peut-être pas un projet Node/Remotion valide.")

# Affiche le chemin actuel pour vérifier
print(f"\n📍 Répertoire de travail actuel : {os.getcwd()}")
```
